```python
# Load pipeline config
import json
from pathlib import Path
cfg_path = Path('config.json')
if cfg_path.exists():
    with cfg_path.open('r', encoding='utf-8') as _f:
        _cfg = json.load(_f)
else:
    _cfg = {}
DATA_PATH = _cfg.get('DATA_PATH', 'data/EEG_data.csv')
RANDOM_STATE = _cfg.get('RANDOM_STATE', 42)
N_SPLITS = _cfg.get('N_SPLITS', 10)
TOP_FEATURES = _cfg.get('TOP_FEATURES', 20)
print('Loaded config:', DATA_PATH, RANDOM_STATE, N_SPLITS, TOP_FEATURES)
```
